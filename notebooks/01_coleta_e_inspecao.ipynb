## Diagnóstico de qualidade dos dados

Base: Inside Airbnb, Rio de Janeiro, download em 24/11/2026. 48.247 anúncios, 90 colunas originais, 21 selecionadas.

| Problema | Causa provável | Decisão |
|---|---|---|
| `neighbourhood` 100% nula | Sem descrição no dicionário de dados; provavelmente não coletada nesta extração. | Usar `neighbourhood_cleansed`, que tem o bairro padronizado. |
| `bathrooms` 100% nula | O dicionário a define como numérica, mas veio vazia; a informação está em `bathrooms_text` (ex: "2 baths"). | Usar `bathrooms_text` e extrair o número na etapa 02. |
| `price` como texto | Contém caracteres não numéricos ("$" e pontuação de milhar). | Remover "$" e pontuação e converter para número. Decidir entre `price` e `price_quote_price_per_night` na etapa 02. |
| `review_scores_rating` com ~19% nulos | Todos os 9.058 anúncios sem nota têm number_of_reviews = 0: a falta de nota vem da falta de avaliações, não de erro no dado. | Manter os nulos e criar a coluna `tem_avaliacao` (True/False). Preencher com 0 ou com a média distorceria a análise, já que não ter avaliações pode estar ligado ao preço e à nota. |

**Moeda:** apesar do "$", os preços estão em reais. Validei comparando com anúncios atuais no site do Airbnb.

**Disponibilidade:** usei `availability_30` porque as reservas costumam ser feitas poucas semanas antes da estadia, então os próximos 30 dias refletem melhor a ocupação real. Comparando a mediana em porcentagem de dias livres, as janelas de 30, 60 e 90 dias ficam em 70%, 78% e 80%, uma tendência coerente (quanto mais distante, menos reservas). Já a de 365 dias cai para 49%, provavelmente porque muitos anfitriões só abrem a agenda alguns meses à frente, e os dias bloqueados contam como indisponíveis. Por isso a `availability_365` mistura reservas com agenda fechada e não serve como medida de ocupação.
